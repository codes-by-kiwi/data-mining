```python
from scipy.sparse import coo_matrix
import pandas as pd
import numpy as np

# data = np.load('/Users/fizzausman/Downloads/user_movie_rating (1).npy')
# # user_id, movie_id and rating

# df = pd.DataFrame(data, columns=['user_id', 'movie_id', 'rating'])

user_ids = df['user_id'].values #e.g. array([510, 611, 612, 620, 614, 615, 616, 617, 645])
movie_ids = df['movie_id'].values #e.g. array([28, 8, 30, 175, 200, 17692, 17692, 17698, 17695])
ratings = df['rating'].values# e.g. array([1,2,3,4,5])

#sparse matrix 
sparse_matrix = coo_matrix((ratings, (user_ids, movie_ids)))
print("Sparse matrix starts here")
print(sparse_matrix)
print("Sparse matrix ends here")

sparse_matrix_csr = sparse_matrix.tocsr()
print("Sparse matrix in csr format starts here")
print(sparse_matrix_csr) # CSR format is ideal for row slicing
print("Sparse matrix in csr format ends here")
# Outputs this:
# Coords	Values
#   (1, 30)	3
#   (1, 157)	3
#   (1, 173)	4
#   (1, 175)	5
#   (1, 191)	2

print(f"Sparse matrix dimensions: {sparse_matrix.shape}") # will be same as that of csr 
print(f"Sparse matrix in csr dimensions: {sparse_matrix.shape}")
print(f"Number of non-zero elements in sparse_matrix: {sparse_matrix.nnz}")# will be same as that of csr 
print(f"Number of non-zero elements in csr_sparse_matrix: {sparse_matrix_csr.nnz}")


# If you want to inspect a small portion of the matrix
print(sparse_matrix.toarray()[630:635])
print(sparse_matrix_csr.toarray()[630:635])
```

```python
# min hash function

interaction_history = sparse_matrix.copy()
interaction_history.data = np.ones_like(interaction_history.data)  # set all ratings to 1
#print(f"Interaction matrix dimensions: {interaction_history.shape}") # Interaction matrix dimensions: (103704, 17771)
interaction_history_dense = interaction_history.toarray()
interaction_history_dense = interaction_history_dense[1:, :]
print(interaction_history_dense[:5, :10]) 
```

```python
num_permutations = 20

num_movies = sparse_matrix.shape[1]

permutations = [np.random.permutation(num_movies) for _ in range(num_permutations)]

print(permutations)

print("Printing the first permutation across movies")
print(permutations[0])
print("Printing the second permutation across movies")
print(permutations[1])
```

```python
unique_user_count = df.iloc[:, 0].nunique()

num_permutations = 100
num_users = unique_user_count

inf_initial_matrix = np.full((num_permutations, num_users), np.nan) # initial signature matrix
print(inf_initial_matrix)
print(inf_initial_matrix.shape)

# rows represent permutations
# cols represent the users
# values in a cell represent minimum index of rated movie after permutation is applied
```

```python
# signature_matrix = []

# for permutation in num_permutations:
#     signature_matrix

# dense representation of interaction history is :

# [[1, 0, 0, 1], user 0 rated movies 0 and movie 3
#  [0, 1, 1, 0],
#  [0, 0, 0, 1]]

# example permutations:

# [
#     [2, 0, 3, 1],  -> current perm
#     [3, 1, 0, 2],  
#     [0, 3, 2, 1]   
# ]
# rated movies = [0,3]
# for [0,3] permutation will map 0 as movie 2 and 3 as movie 1, permuted_rated_movies = [2,1]

print(interaction_history_dense)
print(interaction_history_dense.shape)

for perm_idx, perm in enumerate(permutations): 
    for user_idx in range(num_users):
        
        rated_movies = np.where(interaction_history_dense[user_idx, :] == 1)[0]  #indexed by zero as np.where returns
        
        if len(rated_movies) > 0:
            permuted_rated_movies = perm[rated_movies] # in specific permutation, you find the movie ids corresponding to the rated movies of 1 
        else:
            permuted_rated_movies = []
        
        if len(permuted_rated_movies) > 0:
            inf_initial_matrix[perm_idx, user_idx] = np.min(permuted_rated_movies)  

        else:
            print(f"User {user_idx} has no rated movies after permutation {perm_idx}")

print("Signature Matrix:")
print(inf_initial_matrix)
print("The shape is:")
print(inf_initial_matrix.shape)
```
